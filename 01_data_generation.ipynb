# Credit Card Fraud Detection - Data Generation

This notebook demonstrates how to generate synthetic transaction data for fraud detection training.

## Why Synthetic Data?

1. **Privacy**: Real transaction data contains sensitive PII (Personally Identifiable Information)
2. **Accessibility**: Students can run this without access to banking systems
3. **Control**: We can adjust fraud rates, patterns, and data volume
4. **Safety**: No risk of exposing real customer data

## Data Schema

Our synthetic dataset includes:
- Transaction identifiers (anonymized)
- Temporal features (hour, day, timestamp)
- Monetary features (amount, velocity)
- Behavioral features (device, channel, location)
- Historical features (24h/1h aggregates)
- Target variable (is_fraud)

```python
# Setup
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import our data generator
from src.generate_synthetic_data import generate_dataset, save_dataset

print("Libraries imported successfully!")
```

## Generate Dataset

We'll generate 100,000 transactions with a 1.5% fraud rate (typical for credit cards).

```python
# Generate synthetic data
df = generate_dataset(n_samples=100000, fraud_ratio=0.015)

# Display first few rows
df.head(10)
```

## Data Statistics

```python
# Basic statistics
print("Dataset Statistics:")
print("=" * 50)
print(f"Total transactions: {len(df):,}")
print(f"Fraudulent transactions: {df['is_fraud'].sum():,}")
print(f"Normal transactions: {(df['is_fraud'] == 0).sum():,}")
print(f"Fraud rate: {df['is_fraud'].mean():.2%}")
print(f"\nDate range: {df['ts'].min()} to {df['ts'].max()}")
```

```python
# Amount statistics by class
print("\nAmount Statistics by Class:")
print(df.groupby('is_fraud')['amount'].describe())
```

## Visualize Data

```python
# Fraud distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Pie chart
fraud_counts = df['is_fraud'].value_counts()
labels = ['Normal', 'Fraud']
colors = ['#2ecc71', '#e74c3c']
axes[0].pie(fraud_counts, labels=labels, autopct='%1.2f%%', 
          colors=colors, startangle=90, explode=(0, 0.1))
axes[0].set_title('Transaction Distribution')

# Bar chart
axes[1].bar(labels, fraud_counts.values, color=colors)
axes[1].set_ylabel('Count (log scale)')
axes[1].set_title('Transaction Count by Class')
axes[1].set_yscale('log')

plt.tight_layout()
plt.show()
```

```python
# Amount distribution
plt.figure(figsize=(10, 6))

normal = df[df['is_fraud'] == 0]['amount']
fraud = df[df['is_fraud'] == 1]['amount']

plt.hist(normal, bins=50, alpha=0.7, label='Normal', 
         color='green', density=True, range=(0, 1000))
plt.hist(fraud, bins=50, alpha=0.7, label='Fraud', 
         color='red', density=True, range=(0, 1000))

plt.xlabel('Transaction Amount')
plt.ylabel('Density')
plt.title('Transaction Amount Distribution')
plt.legend()
plt.show()
```

## Save Dataset

```python
# Save to disk
csv_path, parquet_path = save_dataset(df, output_dir='../data')

print(f"\nDataset saved:")
print(f"  CSV: {csv_path}")
print(f"  Parquet: {parquet_path}")
```
